```python
driver = crear_driver()
abrir_web(driver)
```

```python
def iniciar_sesion(driver, username, password):

    try:
        WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.NAME, "cedula"))
        )

        username_input = driver.find_element(By.NAME, "cedula")
        username_input.click()
        username_input.send_keys(username)
        time.sleep(2)

        password_input = driver.find_element(By.NAME, "password")
        password_input.click()
        password_input.send_keys(password)
        time.sleep(2)

        #boton_login = driver.find_element(By.XPATH, '//*[@id="loginForm"]/div/div[3]/button')
        boton_login = driver.find_element(By.ID, "loginButton")
        boton_login.click()
        print(" Intentando iniciar sesión...")
        time.sleep(8)  # Esperar carga post-login
    except Exception as e:
        print(f" Error durante el login: {e}")
```

```python
iniciar_sesion(driver, "123", "123")
```
